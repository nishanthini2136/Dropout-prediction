```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
roadmap = pd.read_csv("../Dataset/personalized_learning_roadmap.csv")

roadmap.head()
```

```python
print("="*60)

print("Dataset Shape :", roadmap.shape)

print()

print(roadmap.info())

print("="*60)
```

```python
current = roadmap["Risk Level"].value_counts()

print(current)

current.plot(
    kind="bar",
    figsize=(6,4),
    title="Current Student Risk Distribution"
)

plt.ylabel("Students")
plt.show()
```

```python
future = roadmap["Future Risk"].value_counts()

print(future)

future.plot(
    kind="bar",
    figsize=(6,4),
    title="Weekly Future Risk Distribution"
)

plt.ylabel("Students")
plt.show()
```

```python
intervention = roadmap["Recommended Intervention"].value_counts()

print(intervention)

intervention.plot(
    kind="barh",
    figsize=(8,5),
    title="Recommended Interventions"
)

plt.show()
```
