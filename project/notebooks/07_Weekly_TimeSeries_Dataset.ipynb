```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
studentVle = pd.read_csv("../Dataset/processed/studentVle_clean.csv")

studentAssessment = pd.read_csv("../Dataset/processed/studentAssessment_clean.csv")

assessments = pd.read_csv("../Dataset/processed/assessments_clean.csv")

studentInfo = pd.read_csv("../Dataset/processed/studentInfo_clean.csv")
```

```python
print(studentVle.shape)
print(studentAssessment.shape)
print(assessments.shape)
print(studentInfo.shape)
```

```python
studentVle["week"] = ((studentVle["date"] - studentVle["date"].min()) // 7) + 1
```

```python
weekly_clicks = (
    studentVle
    .groupby(
        ["id_student", "code_module", "code_presentation", "week"]
    )["sum_click"]
    .sum()
    .reset_index()
)

weekly_clicks.rename(
    columns={"sum_click": "weekly_video_clicks"},
    inplace=True
)

weekly_clicks.head()
```

```python
weekly_login = (
    studentVle
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]
    )["date"]
    .nunique()
    .reset_index()
)

weekly_login.rename(
    columns={
        "date":"weekly_login_frequency"
    },
    inplace=True
)

weekly_login.head()
```

```python
weekly_activity = (
    studentVle
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]
    )["date"]
    .mean()
    .reset_index()
)

weekly_activity.rename(
    columns={
        "date":"weekly_avg_activity_day"
    },
    inplace=True
)

weekly_activity.head()
```

```python
assessment_data = pd.merge(
    studentAssessment,
    assessments,
    on="id_assessment",
    how="left"
)

assessment_data.head()
```

```python
assessment_data["week"] = (
    (assessment_data["date_submitted"] - assessment_data["date_submitted"].min()) // 7
) + 1
```

```python
weekly_quiz = (

    assessment_data

    .groupby(

        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]

    )["score"]

    .mean()

    .reset_index()

)

weekly_quiz.rename(

    columns={

        "score":"weekly_avg_quiz_score"

    },

    inplace=True

)

weekly_quiz.head()
```

```python
weekly_submission = (

    assessment_data

    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]
    )["date_submitted"]

    .mean()

    .reset_index()

)

weekly_submission.rename(
    columns={
        "date_submitted":"weekly_avg_submission_day"
    },
    inplace=True
)

weekly_submission.head()
```

```python
weekly_weight = (

    assessment_data

    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]
    )["weight"]

    .mean()

    .reset_index()

)

weekly_weight.rename(
    columns={
        "weight":"weekly_avg_assessment_weight"
    },
    inplace=True
)

weekly_weight.head()
```

```python
weekly_assessment = (

    assessment_data

    .groupby(

        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]

    )

    .size()

    .reset_index(name="weekly_assessments_completed")

)

weekly_assessment.head()
```

```python
weekly_data = pd.merge(
    weekly_clicks,
    weekly_login,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)

weekly_data = pd.merge(
    weekly_data,
    weekly_activity,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)

weekly_data = pd.merge(
    weekly_data,
    weekly_quiz,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)

weekly_data = pd.merge(
    weekly_data,
    weekly_assessment,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)

weekly_data = pd.merge(
    weekly_data,
    weekly_submission,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)

weekly_data = pd.merge(
    weekly_data,
    weekly_weight,
    on=["id_student","code_module","code_presentation","week"],
    how="left"
)
```

```python
weekly_data = pd.merge(

    weekly_data,

    studentInfo[

        [

            "id_student",

            "code_module",

            "code_presentation",

            "dropout"

        ]

    ],

    on=[

        "id_student",

        "code_module",

        "code_presentation"

    ],

    how="left"

)
```

```python
weekly_data.fillna({

    "weekly_avg_quiz_score":0,

    "weekly_assessments_completed":0,

    "weekly_login_frequency":0,

    "weekly_avg_activity_day":0,

    "weekly_avg_submission_day":0,

    "weekly_avg_assessment_weight":0

}, inplace=True)
```

```python
weekly_data = weekly_data.sort_values(

    [

        "id_student",

        "week"

    ]

)

weekly_data.head()
```
