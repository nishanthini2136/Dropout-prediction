```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
studentVle = pd.read_csv("../Dataset/processed/studentVle_clean.csv")

studentAssessment = pd.read_csv("../Dataset/processed/studentAssessment_clean.csv")

assessments = pd.read_csv("../Dataset/processed/assessments_clean.csv")

studentInfo = pd.read_csv("../Dataset/processed/studentInfo_clean.csv")
```

```python
print(studentVle.shape)
print(studentAssessment.shape)
print(assessments.shape)
print(studentInfo.shape)
```

```python
studentVle["week"] = (studentVle["date"] // 7) + 1

studentVle.head()
```

```python
weekly_clicks = (

    studentVle

    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]
    )["sum_click"]

    .sum()

    .reset_index()

)

weekly_clicks.rename(

    columns={
        "sum_click":"weekly_video_clicks"
    },

    inplace=True

)

weekly_clicks.head()
```

```python
assessment_data = pd.merge(

    studentAssessment,

    assessments,

    on="id_assessment",

    how="left"

)
```

```python
assessment_data["week"] = (

    assessment_data["date_submitted"] // 7

) + 1
```

```python
weekly_quiz = (

    assessment_data

    .groupby(

        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]

    )["score"]

    .mean()

    .reset_index()

)

weekly_quiz.rename(

    columns={

        "score":"weekly_avg_quiz_score"

    },

    inplace=True

)

weekly_quiz.head()
```

```python
weekly_assessment = (

    assessment_data

    .groupby(

        [
            "id_student",
            "code_module",
            "code_presentation",
            "week"
        ]

    )

    .size()

    .reset_index(name="weekly_assessments_completed")

)

weekly_assessment.head()
```

```python
weekly_data = pd.merge(

    weekly_clicks,

    weekly_quiz,

    on=[

        "id_student",

        "code_module",

        "code_presentation",

        "week"

    ],

    how="left"

)

weekly_data = pd.merge(

    weekly_data,

    weekly_assessment,

    on=[

        "id_student",

        "code_module",

        "code_presentation",

        "week"

    ],

    how="left"

)
```

```python
weekly_data = pd.merge(

    weekly_data,

    studentInfo[

        [

            "id_student",

            "code_module",

            "code_presentation",

            "dropout"

        ]

    ],

    on=[

        "id_student",

        "code_module",

        "code_presentation"

    ],

    how="left"

)
```

```python
weekly_data.fillna(

    {

        "weekly_avg_quiz_score":0,

        "weekly_assessments_completed":0

    },

    inplace=True

)
```

```python
weekly_data = weekly_data.sort_values(

    [

        "id_student",

        "week"

    ]

)

weekly_data.head()
```

```python
print(weekly_data.shape)

weekly_data.head()
```

```python
weekly_data.to_csv(

    "../Dataset/weekly_student_dataset.csv",

    index=False

)

print("Weekly Student Dataset Saved Successfully")
```
